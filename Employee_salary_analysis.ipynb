```python
# Employee Salary Analysis
Yeh notebook employee salaries ko analyze karne ke liye banai gayi hai[span_1](start_span)[span_1](end_span). Ismein manually created list data structure aur Python ke built-in functions ka use kiya gaya hai[span_2](start_span)[span_2](end_span).
```

```python
# Manually created employee salary data stored in a list
salaries = [45000, 52000, 38000, 60000, 41000, 75000, 32000]

print("Original Salary Data:", salaries)
```

```python
# Using built-in Python functions for basic calculations
total_salary = sum(salaries)
highest_salary = max(salaries)
lowest_salary = min(salaries)
number_of_employees = len(salaries)

# Calculating average salary
average_salary = total_salary / number_of_employees

# Salary statistics and results
print("=== Salary Analysis Results ===")
print(f"Total Employees: {number_of_employees}")
print(f"Total Salary Payout: ₹{total_salary}")
print(f"Average Salary: ₹{average_salary:.2f}")
print(f"Highest Salary: ₹{highest_salary}")
print(f"Lowest Salary: ₹{lowest_salary}")
```
